```python
import pandas as pd
import yfinance as yf
from google.colab import files # Importamos la herramienta de Colab para descargar

# 1. Descargar y preparar los datos
assets_prueba = ['^STOXX50E']
data = yf.download(assets_prueba, start="2022-01-01", end="2025-12-13", auto_adjust=False)
data.reset_index(inplace=True)

if isinstance(data.columns, pd.MultiIndex):
    data.columns = [col[0] for col in data.columns]

data = data[['Date', 'Open', 'High', 'Low', 'Close']].copy()
data['Date'] = pd.to_datetime(data['Date']).dt.strftime('%m/%d/%Y')

# 2. Guardar el archivo en el almacenamiento temporal de Colab
ruta_archivo = 'serie_eurostoxx50.csv'
data.to_csv(ruta_archivo, index=False)

# 3. Forzar la descarga a tu ordenador físico
files.download(ruta_archivo)

print("Descarga completada.")
```
