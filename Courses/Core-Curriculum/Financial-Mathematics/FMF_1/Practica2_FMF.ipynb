## Ejercicios

a) Supongamos que el factor de crecimiento de una cuenta bancaria viene dado por $B_{0}=1$, $B_{T}=1.12$. Si $S_{0}=34$ euros, ¿es posible encontrar una oportunidad de arbitraje si se establece el precio forward de $S$ con expedición en $T$ en $38,60$ euros?

b) Si $B_{T}=1.05$, el precio de hoy de una libra es $L_{0}=1.60$ euros y el tipo de cambio forward para tiempo de expedición $T$ es $K_0=1.50$ euros por cada libra. ¿Cuánto vale a día de hoy un bono que paga 100 libras a tiempo $T$?

```python
# a) Calculamos el precio forward que debería tener el activo
S0 = 34
BT = 1.12
K0 = S0*BT # 38.08
```

El precio forward calculado (que es el correcto) resulta ser menor que el que se establece en el mercado. Esto genera una oportunidad de arbitraje. El mercado nos permite entrar en un forward de precio nulo con $K=38.60$ euros. Si construimos una cartera con una unidad del activo $S$ pidiendo $K_0/B_T$ al banco, esta cartera nos cuesta $0$ y al final del intervalo valdrá
$$S_T-K_0$$
Por otra parte, el forward que ofrece el mercado a tiempo $T$ paga:
$$S_T-K<S_T-K_0$$ en todos los escenarios futuros. Esto quiere decir que conviene entrar como *vendedor del forward del mercado*, y al mismo tiempo construir la cartera indicada. Los valores finales de nuestra inversión son siempre positivos, y el valor inicial es nulo:
$$(S_T-K_0)-(S_T-K) = K-K_0>0$$ en todo escenario futuro.

```python
# b) El tipo de cambio forward viene dado por K0 = L0*ZL*BT
# donde L0 es el precio actual de la libra,
# ZL es el precio de un bono cupón cero en libras,
# y BT es el crecimiento bancario
L0 = 1.6
K0 = 1.5
BT = 1.05
ZL = K0/(L0*BT) # 0.8928571428571428 en libras
ZLe = K0/BT # 1.4285714285714286 en euros
# éste es el precio del bono que paga una unidad de libra a tiempo T,
# si paga 100 se multiplica por 100
```
