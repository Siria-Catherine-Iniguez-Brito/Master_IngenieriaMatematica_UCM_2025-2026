## Ejercicio 1

Considera un activo $S$ que evoluciona según un modelo Browniano geométrico:  
$$ \frac{\Delta S_n}{S_{n-1}} = \mu \tau + \sigma \sqrt{\tau} Z_n $$ donde
$$ Z_n=\begin{cases} +1& p\\
-1& 1-p\end{cases}$$

Hemos visto que para $\tau\to 0$ este modelo proporciona valores finales que tienen una distribución lognormal en la medida neutral al riesgo. Calcula el precio del contrato que paga a tiempo $T$:
$$
f\left(S_T-S_0\right)
$$
donde
$$
f(x)=\begin{cases}
0 & x < 0\\
1-\exp(-x) & x \geq 0
\end{cases}
$$
para $T=1$, $\sigma = 0.25$, $r=8\%$ (tipo continuo), $S_0=100$.  
**Indicación**: Se puede calcular el precio de tres formas:
- Con el árbol binomial adecuado para $\tau$ pequeño.
- Con la distribución continua final de los precios en la medida neutral al riesgo.
- Con un método tipo Montecarlo, es decir, generando valores finales de $S_T$ con la distribución adecuada y estimando el valor esperado de la fórmula de valoración con el promedio.

## Ejercicio 2

Considera una opción de compra con strike $K=100$ sobre una acción $S$ con $S_{0}=90$. El tiempo de ejercicio es de 6 meses (T=1/2). El modelo de precios de la acción viene dado por
$$\frac{\Delta S_{n}}{S_{n-1}}=0.25\tau+0.20\sqrt{\tau}Z_n $$ donde $Z_n$ tiene la misma distribución del **Ejercicio 1** con $p=1/2$. El tipo de interés continuo se asume constante y viene dado por $r=6\%$ anual.

Realiza $N_{\text{sim}}=10^{4}$ simulaciones de los caminos del activo $S$ para distintos pasos temporales $\tau=T/N$ con: $N=10^{j}$ y $j=2,3$. Para cada camino, utiliza la estrategia de replicación de la opción descrita en la teoría, calculando el precio inicial de la opción mediante la fórmula de Black-Scholes. Dibuja un histograma para cada valor de $j$ con los valores de la diferencia entre el precio de la cartera de replicación y el valor de la opción a tiempo $T$.
