## Ejercicio 1: Valor presente de una sucesión de flujos de caja

```python
x = np.array([20,25,30,35,40,30,20,10])
Z = np.array([1,0.943,0.935,0.93,0.926,0.923,0.921,0.917]) # incluimos el factor 1 para el pago actual
```

Para determinar el valor presente neto en los distintos instantes de tiempo, hay que calcular los factores de descuento que corresponden a cada fecha intermedia. Empezamos con $t=T_0$

```python
VP = np.zeros(8) # inicializamos el vector de valores presentes
VP[0] = np.dot(x,Z)
```

```python
VP
```

Para las fechas sucesivas, hay que tener en cuenta que los pagos ya realizados no entran en el cálculo, y hay que usar los factores de descuento a partir del día actual, por ejemplo, para el día 1:

```python
VP[1] = 1/Z[1]*np.dot(x[1:],Z[1:])
```

El esquema se repite, de manera que lo ponemos todo en un bucle:

```python
for i in range(1,8):
    VP[i] = 1/Z[i]*np.dot(x[i:],Z[i:])
```

```python
VP
```

## Ejercicio 2: Valor presente en función de los tipos de interés

Definimos los flujos, los plazos y los tipos de interés:

```python
x = np.array([-40,10,10,10,10,10,10])
pl = np.arange(1,7)
R  = np.array([5.0,5.3,5.6,5.8,6.0,6.1])/100 # a partir del plazo de 1 año
```

Definimos ahora los factores de descuento en función del interés y de los plazos, agregamos el factor de descuento correspondiente a $t=0$, que siempre es 1:

```python
Da = np.zeros(7) # descuentos para composición anual
Da[0] = 1
Da[1:7] = 1/(1 + R)**pl
```

Aplicamos ahora lo visto en el ejercicio 1:

```python
VP = np.zeros(7)
for i in range(7):
    VP[i] = 1/Da[i]*np.dot(x[i:7],Da[i:7])
```

```python
VP
```
